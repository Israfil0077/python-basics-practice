### Predict the value of Y (price) based on the value of X (size of the apartment in square feet).
       X(size in sq ft)           Y(Price in Usd)
             3                          180
             4                          360
             5                          540
             10                         1440
             12                         1800

### Given the size (X) of an apartment, use this dataset to predict the corresponding price (Y). For example, what would be the expected price of an apartment that is 7 sq ft?

## Step-1 Training Dataset

```python
import numpy as np

X_train =np.array([3,4,5,10,12])
Y_train = np.array([180,360,540,1440,1800])
```

```python
import matplotlib.pyplot as plt

plt.scatter(X_train,Y_train)
plt.title("Training Data")
plt.xlabel("X_train")
plt.ylabel("Y_train")
plt.grid()
plt.show()
```

# Initial Prediction

```python
w1=0
w0=0

y_predicted = w1*X_train + w0

plt.scatter(X_train,Y_train, color='blue', label='Ground Truth')

plt.plot(X_train,y_predicted, color='red', label='Initial Predictions')

plt.title("Training Data + Initial Predictions")
plt.xlabel("X_train")
plt.ylabel("Y_train")
plt.legend()
plt.grid(True)
plt.show()
```

# Gradient Descent Algorithm

### Gradient Descent is an optimization algorithm used to minimize the loss function 
### (in this case, Mean Squared Error)

### The algorithm iteratively adjusts the model’s parameters to find the values that minimize the errors between predicted and actual values.

### How it works:
### Start with random initial values for the model parameters (weights).
### Compute the gradient of the loss function, which tells you the direction of the steepest ascent.
### Update the weights in the opposite direction of the gradient to reduce the error. 
### Repeat until the loss function converges (reaches a minimum).

```python
learning_rate = 0.001
epochs = 1000

w1 =0
w0 =0

error_history = []

for epoch in range(epochs):

    total_epoch_error = 0

    for i in range(len(X_train)):
        X_sample = X_train[i]
        Y_sample = Y_train[i]
        y_predicted = w1*X_sample + w0

        delta_w1 = -(Y_sample - y_predicted) * X_sample
        delta_w0 = -(Y_sample - y_predicted)

        w1 = w1 - learning_rate * delta_w1
        w0 = w0 - learning_rate * delta_w0

        total_epoch_error += (Y_sample - y_predicted) ** 2

    error_history.append(total_epoch_error / len(X_train))

print(f"w1 = {w1:.4f}, w0 = {w0:.4f}")
print(f"\nPredicted y = {w1:.4f}*X + {w0:.4f}")
```

```python
y_predicted = w1 * X_train + w0

plt.scatter(X_train, Y_train, color='blue', label='Ground truths')

plt.plot(X_train, y_predicted, color='red', label='Initial predictions')

plt.title("Training Data + Initial Predictions")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```

# Error Over Time

```python
plt.figure(figsize=(8,6))
plt.plot(range(epochs), error_history)

plt.title("Model Error Over Time")
plt.xlabel("Epochs")
plt.ylabel("Loss")

plt.grid(True)
plt.show()
```
